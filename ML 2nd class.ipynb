```python
x= np.array([1,2,3,4])
print("Original Array:")
print(x)
print("test if none of the element of the said array is zero :")
print(np.all(x))
x=np.array([0,1,2,3])
print("original array:")
print(x)
print("test if none of the element of the said array is zero :")
print(np.all(x))
```
